
***DEFINE THE DATAFRAME***

First, we will define a DataFrame with pandas that contains all the dictionary keys as columns, so that we can operate and analize the data easily later on.

```python
# orient='index' makes each dictionary key a row index.
all_printers_df = pd.DataFrame.from_dict(all_printers.printers_data, orient='index')

# The hostnames are now the DataFrame's "index". Let's make them a regular column named 'Hostname'.
all_printers_df.reset_index(inplace=True)
all_printers_df.rename(columns={'index': 'Hostname'}, inplace=True)

#all_printers_df.head()
```

We will also add a column called 'Base Code' to group printers by base later

```python
all_printers_df['Base Code'] = all_printers_df['Hostname'].str[:4]
```

***GET A COUNT OF TOTAL ONLINE AND OFFLINE PRINTERS***

We will start by grouping the printers by status

```python
# A simple dictionary to hold the ANSI codes for formatting
colors = {
    "red": "\033[91m",
    "green": "\033[92m",
    "bold": "\033[1m",
    "end": "\033[0m"  # This code resets all formatting
}

# Group rows by only the 'base_code' (site)
grouped_by_site = all_printers_df.groupby('Base Code')

print("\n--- Printer Status by Site ---")

# The loop now yields the site name and its corresponding data
for site_code, group_data in grouped_by_site:
    
    # Print the site header in BOLD
    header = f"\n--- Site: {site_code} ---"
    print(f"{colors['bold']}{header}{colors['end']}")

    if group_data.empty:
        continue

    # Iterate through each printer in the group to apply custom formatting
    for index, row in group_data.iterrows():
        hostname = row['Hostname']
        status = row['Status']
        
        formatted_status = status  # Default to plain text
        
        # Apply color and bold formatting based on the status
        if status == 'Online':
            formatted_status = f"{colors['bold']}{colors['green']}{status}{colors['end']}"
        elif status == 'Offline':
            formatted_status = f"{colors['bold']}{colors['red']}{status}{colors['end']}"

        # Print the formatted row. The padding helps align the columns.
        print(f"{hostname:<25} {formatted_status}")
```

Lets give the counts in text format...

```python
# Get the count of each status
status_counts =  all_printers_df.groupby('Status')

# Display the text counts
print(status_counts.size())
```
